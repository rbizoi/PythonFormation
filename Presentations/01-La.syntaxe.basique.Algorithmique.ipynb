# LIRE en Python

```python
entree = input('Veuillez entrer un entier') # entree est une string
print(entree, type(entree)) 	# <class str>
entree = int(entree) # entree est un entier, eval(entree) pour être plus générique
print(entree, type(entree)) 	# <class 'int'>
```

# Opérateurs de comparaison

# Opérateurs logiques

```python
print(True and False) 	# affiche False 
print(True and True) 	# affiche True 
print(True or False) 	# affiche True 
print(1 == 1 or False) 	# affiche True
print(not False) 		# affiche True
```
