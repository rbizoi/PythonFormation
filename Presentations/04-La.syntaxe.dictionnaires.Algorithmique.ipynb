# Parcours

```python
mon_dico = {}	
mon_dico['computer'] = 'ordinateur'
mon_dico['mouse'] = 'souris'
mon_dico['keyboard'] = 'clavier'
```

## clé simple

```python
for key in mon_dico :
	print(key, ":", mon_dico[key])
```

## avec l'opération keys()

```python
for key in mon_dico.keys() :
	print(key, ":", mon_dico[key])
```

## avec l'opération items()

```python
for key, value in mon_dico.items() :
	print(key, ":", value)
```
