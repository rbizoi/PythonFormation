# Opérateur ternaire

```python
age = int(input("Entrer votre âge"))
if age < 18: res = 'mineur'
else : res = 'majeur'
print("Vous êtes", res)
```

```python
age = int(input("Entrer votre âge"))
res = 'mineur' if age < 18 else 'majeur'
print("Vous êtes", res)
```
