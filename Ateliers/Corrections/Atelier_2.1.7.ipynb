Écrivez un programme qui calcule le signe du produit de deux réels entrés par l’utilisateur sans utiliser la multiplication ni aucun autre calcul.

```python
a = eval(input("entrer un nombre "))
b = eval(input("entrer un nombre "))

print("nulle" if a == 0 or b == 0 else "positif" if a > 0 and b > 0 else "negatif")
```
