# Tri par sélection

Le tri par sélection recherche le plus petit élément de la partie non triée du tableau, puis le place à la bonne position.
>> [7, 4, 9, 2, 5]<br>
>> Minimum = 2<br>
>> [2, 4, 9, 7, 5]<br>
>> Minimum restant = 4<br>
>> [2, 4, 9, 7, 5]<br>
>> Minimum restant = 5<br>
>> [2, 4, 5, 7, 9]<br>


>>```
>>    ALGORITHME TriSelection
>>    
>>    VAR
>>        tab : TABLEAU
>>        i, j, indiceMin, temp : ENTIER
>>    DEBUT
>>        POUR i <- 0 A longueur(tab) - 2 FAIRE
>>            indiceMin <- i
>>            POUR j <- i + 1 A longueur(tab) - 1 FAIRE
>>                SI tab[j] < tab[indiceMin] ALORS
>>                    indiceMin <- j
>>                FINSI
>>            FINPOUR
>>            temp <- tab[i]
>>            tab[i] <- tab[indiceMin]
>>            tab[indiceMin] <- temp
>>        FINPOUR
>>    FIN
>>```

````python
def tri_selection(tab):
    n = len(tab)
    for i in range(n - 1):
        indice_min = i
        for j in range(i + 1, n):
            if tab[j] < tab[indice_min]:
                indice_min = j
        tab[i], tab[indice_min] = tab[indice_min], tab[i]
    return tab


liste = [9, 12, 40, 5, 12, 3, 27, 5, 9, 3, 8, 22, 40, 3, 2, 4, 6, 25]
print("Avant :", liste)
tri_selection(liste)
print("Après :", liste)
````

# Tri par insertion

Le tri par insertion construit progressivement une partie triée du tableau. Le principe ressemble au rangement de cartes à jouer dans sa main.

>> [7, 4, 9, 2, 5]<br>
>> 7 est déjà trié<br>
>> Insertion de 4 :<br>
>> [4, 7, 9, 2, 5]<br>
>> Insertion de 9 :<br>
>> [4, 7, 9, 2, 5]<br>
>> Insertion de 2 :<br>
>> [2, 4, 7, 9, 5]<br>
>> Insertion de 5 :<br>
>> [2, 4, 5, 7, 9]<br>
>> ```
>>     ALGORITHME TriInsertion
>>     VAR
>>         tab : TABLEAU
>>         i, j, valeur : ENTIER
>>     DEBUT
>>         POUR i <- 1 A longueur(tab) - 1 FAIRE
>>             valeur <- tab[i]
>>             j <- i - 1
>>             TANT QUE j >= 0 ET tab[j] > valeur FAIRE
>>                 tab[j + 1] <- tab[j]
>>                 j <- j - 1
>>             FINTANTQUE
>>             tab[j + 1] <- valeur
>>         FINPOUR
>>     FIN
>> ```

````python
def tri_insertion(tab):
    n = len(tab)
    for i in range(1, n):
        valeur = tab[i]
        j = i - 1
        while j >= 0 and tab[j] > valeur:
            tab[j + 1] = tab[j]
            j = j - 1
        tab[j + 1] = valeur
    return tab


liste = [9, 12, 40, 5, 12, 3, 27, 5, 9, 3, 8, 22, 40, 3, 2, 4, 6, 25]
print("Avant :", liste)
tri_insertion(liste)
print("Après :", liste)
````
