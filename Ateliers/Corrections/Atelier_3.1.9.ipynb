### 1. 
Écrivez un algorithme qui calcule la moyenne des valeurs d’un tableau d’entiers. 

```python
tt = [32, 5, 12, 8, 3, 75, 2, 15]
moyenne = 0

for i in tt:
	moyenne += i

moyenne /= len(tt)
print("Le moyenne est de", moyenne)
```

### 2. 
Écrivez un algorithme qui calcule le nombre d’occurrences d’une valeur entrée par l’utilisateur dans un tableau d’entiers (le nombre de fois que la valeur apparaît dans le tableau).

```python
mots=["ordinateur","cafe","chocolat","digestion","faim","formation","television","cafe"]
occurence = 0
for mot in mots :
	if mot == "cafe":
		occurence += 1
print(occurence)
# ou
print(mots.count("cafe"))
```

### 3. 
Écrivez un algorithme qui réalise l’inversion des éléments d’un tableau sans utiliser de tableau intermédiaire. 

```python
mots, motsInverse=["ordinateur","cafe","chocolat","digestion","faim","formation","television"],[]
for m in mots:
    motsInverse.insert(0,m)
print(motsInverse)    
```
