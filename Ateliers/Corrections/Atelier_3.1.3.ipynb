Écrivez un programme qui renvoie le nom du énième mois de l’année.

```python
mois = ['Janvier,', 'Février', 'Mars', 'Avril', 'Mai', 
            'Juin', 'Juillet', 'Août', 'Septembre', 
            'Octobre', 'Novembre', 'Décembre']
n = int(input("renvoie le nom du n-ième mois de l'année :"))
if 1 <= n and n <= 12 : 
    print(mois[n -1]) # les indices sont numérotés à partir de zéro
```
