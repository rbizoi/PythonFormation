Écrivez un programme faisant saisir à l’utilisateur trois entiers, i, j et k, et  les triant par ordre croissant et les afficher pour vérifier votre tri.

```python
x = int(input("Enter un entier : "))
y = int(input("Enter un entier : "))
z = int(input("Enter un entier : "))

if x > y:
	x,y = y,x
if y > z :
	y,z = z,y
if x > y:
	x,y = y,x

print(x,y,z)
```
