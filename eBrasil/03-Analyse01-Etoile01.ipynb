# 01 Dimension d'analyse le temps avec la base de travail la date de commande et métrique le chiffre d’affaire

```python
plt.figure(figsize=(32,24))
gbarplot = sns.barplot(x='annee',y='value',data=donnees,palette=palette,alpha=0.8)
xloc, xlabels = plt.xticks()
gbarplot.set_xticklabels(xlabels, rotation=45);
```

**Attention** au estimateur par défaut il prend en compte la moyenne 

```python
plt.figure(figsize=(32,24))
gbarplot = sns.barplot(x='annee',y='value',data=donnees,palette=palette,alpha=0.8,estimator='sum')
xloc, xlabels = plt.xticks()
gbarplot.set_xticklabels(xlabels, rotation=45);
```

 La somme de chiffre d’affaire en 2018 est supérieure au 2017, on va continuer l’analyse sur les mois des deux années.

```python
plt.figure(figsize=(32,24))
gbarplot = sns.barplot(x='annee_mois',y='value',data=donnees[donnees.annee > 2016],palette=palette,alpha=0.8,estimator='sum', hue_order='value')
xloc, xlabels = plt.xticks()
gbarplot.set_xticklabels(xlabels, rotation=45);
```

```python
affichage = donnees[donnees.annee > 2016].groupby('annee_mois').agg({'value':'sum'})
```

```python
affichage
```

```python
plt.figure(figsize=(32,24))
gbarplot = sns.barplot(x='annee_mois',y='value',data=affichage,palette=palette,alpha=0.8)
xloc, xlabels = plt.xticks()
gbarplot.set_xticklabels(xlabels, rotation=45);
```

# Retrouver les cinq mois les plus rentables du point de vues du chiffre d’affaire

```python
affichage = donnees[donnees.annee > 2016].groupby('annee_mois').agg({'value':'sum'})
affichage['r'] = affichage.value.rank(method='dense',ascending=False)
```

```python
affichage[affichage.r < 6]
```

```python
plt.figure(figsize=(32,24))
gbarplot = sns.barplot(x='annee_mois',y='value',data=affichage[affichage.r < 6],palette=palette,alpha=0.8)
xloc, xlabels = plt.xticks()
gbarplot.set_xticklabels(xlabels, rotation=45);
```

```python
lannee_mois = affichage[affichage.r < 6].reset_index().annee_mois
```

```python
lannee_mois
```

```python
affichage = donnees[donnees.annee_mois.isin(lannee_mois)]
```

```python
affichage.pivot_table(index='state',
                      columns='annee_mois',
                      values='value',
                      aggfunc = 'sum')
```

```python
plt.figure(figsize=(24,36))
# sns.set(font_scale=2.4)
sns.heatmap(affichage.pivot_table(index='state',
                      columns='annee_mois',
                      values='value',
                      aggfunc = 'sum'),
            linewidths=0.3, 
            fmt= '.2f', 
            cmap='coolwarm', linecolor='white', annot=True)
# sns.set(font_scale=2)
```
