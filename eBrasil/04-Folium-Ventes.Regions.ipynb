```python
analyse_ventes = pd.read_parquet('etoile03.parquet')
```

```python
donnees = analyse_ventes.loc[(analyse_ventes.annee_mois == 201711)  & 
                             (analyse_ventes.status == 'delivered') & 
                             ( ~ pd.isnull(analyse_ventes.c_lat) ) &
                             ( ~ pd.isnull(analyse_ventes.c_lng) ), ['c_lat', 'c_lng']]
# donnees = donnees[donnees.c_lat < 10]
```

```python
analyse_ventes.columns
```

```python
analyse_ventes.c_name_state.unique()
```

```python
info = analyse_ventes.groupby('c_name_state').agg({'c_lat':'mean',
                                                   'c_lng':'mean', 
                                                   'value_boleto':'sum', 
                                                   'value_credit_card':'sum', 
                                                   'value_debit_card':'sum',
                                                   'value_not_defined':'sum', 
                                                   'value_voucher':'sum'})
# info.columns  = [ nom if nom[:5] == 'value' else nom+'_'+agg for (nom,agg) in info.columns]
info['value'] = info.value_boleto + info.value_credit_card + info.value_debit_card + info.value_voucher + info.value_not_defined
info.drop(columns=['value_boleto', 
 'value_credit_card', 
 'value_debit_card',
 'value_not_defined', 
 'value_voucher'],inplace=True)
info['palmares'] =  info.value.rank(method='dense',ascending=False).astype('int32')
info.reset_index(inplace=True)
info.rename(columns={'c_name_state':'name_state','c_lat':'lat','c_lng':'lng'},inplace=True)
info.head()
```

```python
latitude  = donnees.c_lat.mean()
longitude = donnees.c_lng.mean()

heat_data = [[row['c_lat'],row['c_lng']] for index, row in donnees.iterrows()]
map_brasil = folium.Map(location=[latitude,longitude],zoom_start=4)
plugins.HeatMap(heat_data).add_to(map_brasil)

for index, row in info[info.palmares < 11].iterrows():
    folium.Marker(
        location=[row['lat'],row['lng']],
        tooltip=f"<b>{row['palmares']}</b> - {row['name_state']}",
        popup=f"<b>{row['palmares']}</b> - {row['name_state']}<br>ventes=<b>{row['value']:.0f}</b>",
        icon=folium.Icon(color="blue")).add_to(map_brasil)
    
map_brasil.save("Ventes.Regions.html")

map_brasil
```
